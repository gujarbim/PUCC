## 14. Identificando outliers

Utilize:

```python
Q1 = df["renda"].quantile(0.25)
Q3 = df["renda"].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

outliers = df[
    (df["renda"] < limite_inferior) |
    (df["renda"] > limite_superior)
]

outliers
```

### Questões
